## Learning Goals

## Concept Notes

A perceptron is one of the simplest neural models. It learns a linear decision boundary for binary classification.


## Step-by-Step Example

Run this compact example first. Focus on the inputs, the transformation, and the output.


```python
import numpy as np

X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y = np.array([0, 0, 0, 1])
weights = np.zeros(2)
bias = 0
learning_rate = 0.1

for epoch in range(10):
    for features, target in zip(X, y):
        prediction = 1 if np.dot(features, weights) + bias >= 0.5 else 0
        error = target - prediction
        weights += learning_rate * error * features
        bias += learning_rate * error

print("weights:", weights, "bias:", bias)
print("predictions:", [1 if np.dot(row, weights) + bias >= 0.5 else 0 for row in X])
```
