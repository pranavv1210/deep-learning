# Transfer Learning

A beginner-friendly introduction to transfer learning.


## Transfer learning

Transfer learning starts from a model trained on a large source dataset and adapts it to a related target task. Early layers may already detect useful general patterns, so the target task can require less labelled data and training time.

## Learning Goals

- Understand frozen feature extractors
- Use pretrained models conceptually
- Fine-tune only later layers


### Feature extraction versus fine-tuning

Feature extraction freezes the pretrained backbone and trains a new task-specific head. Fine-tuning unfreezes some or all backbone layers and updates them with a small learning rate. Start with feature extraction when data is limited, then fine-tune carefully if validation performance needs improvement.

## Concept Notes

Transfer learning reuses features learned from a large dataset. Beginners usually freeze the base model first and train a small head.


### Avoid leakage and preprocessing mismatch

Use the preprocessing expected by the pretrained model, including input size and pixel normalization. Split target data before augmentation or model selection. If related images appear across splits, performance can be overly optimistic.

## Step-by-Step Example

Run this compact example first. Focus on the inputs, the transformation, and the output.


## Key terms

- **Pretrained model:** model weights learned on an earlier dataset/task.
- **Backbone:** main feature-extracting part of a network.
- **Head:** task-specific output layers.
- **Freeze:** prevent selected parameters from being updated.
- **Fine-tuning:** continue training pretrained weights on a new task.

```python
try:
    import tensorflow as tf
    base_model = tf.keras.applications.MobileNetV2(
        input_shape=(96, 96, 3), include_top=False, weights=None
    )
    base_model.trainable = False
    model = tf.keras.Sequential([
        base_model,
        tf.keras.layers.GlobalAveragePooling2D(),
        tf.keras.layers.Dense(2, activation="softmax"),
    ])
    model.summary()
except Exception as error:
    print("TensorFlow is not installed. Install requirements.txt to run this notebook fully.")
    print("Environment detail:", error)
```
