# Hyperparameter Tuning

A beginner-friendly introduction to hyperparameter tuning.


## Learning Goals

- Identify common hyperparameters
- Run a small grid search
- Compare validation scores


## Concept Notes

Hyperparameters are choices made before training, such as learning rate, hidden units, depth, and batch size.


## Step-by-Step Example

Run this compact example first. Focus on the inputs, the transformation, and the output.


```python
from sklearn.datasets import make_classification
from sklearn.model_selection import GridSearchCV
from sklearn.neural_network import MLPClassifier

X, y = make_classification(n_samples=300, n_features=8, random_state=42)
param_grid = {
    "hidden_layer_sizes": [(8,), (16,), (16, 8)],
    "alpha": [0.0001, 0.01],
}

search = GridSearchCV(MLPClassifier(max_iter=500, random_state=42), param_grid, cv=3)
search.fit(X, y)
print("Best params:", search.best_params_)
print("Best score:", search.best_score_)
```
