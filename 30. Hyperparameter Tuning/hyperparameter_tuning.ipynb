# Hyperparameter Tuning

A beginner-friendly introduction to hyperparameter tuning.


## Neural-network hyperparameters

Hyperparameters configure the learning process rather than being learned directly from examples. Common choices include learning rate, batch size, network depth, hidden units, dropout rate, optimizer and training epochs.

## Learning Goals

- Identify common hyperparameters
- Run a small grid search
- Compare validation scores


### Search methods

Manual search uses a reasoned sequence of trials. Grid search tests all specified combinations. Random search samples combinations and often explores a wide space more efficiently. Bayesian optimization uses earlier results to select promising future trials.

## Concept Notes

Hyperparameters are choices made before training, such as learning rate, hidden units, depth, and batch size.


### Fair comparisons

Use the same data split and evaluation metric when comparing configurations. Track the random seed and all settings. Tune using validation data, use early stopping where appropriate, and keep the final test set untouched until the model is chosen.

## Step-by-Step Example

Run this compact example first. Focus on the inputs, the transformation, and the output.


## Key terms

- **Hyperparameter:** setting selected outside ordinary gradient learning.
- **Search space:** allowed values for settings.
- **Early stopping:** stop when validation performance ceases improving.
- **Experiment tracking:** recording configuration, metrics, code and artifacts for each trial.

```python
from sklearn.datasets import make_classification
from sklearn.model_selection import GridSearchCV
from sklearn.neural_network import MLPClassifier

X, y = make_classification(n_samples=300, n_features=8, random_state=42)
param_grid = {
    "hidden_layer_sizes": [(8,), (16,), (16, 8)],
    "alpha": [0.0001, 0.01],
}

search = GridSearchCV(MLPClassifier(max_iter=500, random_state=42), param_grid, cv=3)
search.fit(X, y)
print("Best params:", search.best_params_)
print("Best score:", search.best_score_)
```
