## Learning Goals

## Concept Notes

CNNs learn local patterns in images. Early layers detect edges and textures; deeper layers combine them into shapes.


## Step-by-Step Example

Run this compact example first. Focus on the inputs, the transformation, and the output.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import convolve2d

image = np.zeros((8, 8))
image[:, 4:] = 1
edge_filter = np.array([[-1, 1], [-1, 1]])
feature_map = convolve2d(image, edge_filter, mode="valid")

plt.subplot(1, 2, 1)
plt.imshow(image, cmap="gray")
plt.title("Image")
plt.axis("off")
plt.subplot(1, 2, 2)
plt.imshow(feature_map, cmap="gray")
plt.title("Feature map")
plt.axis("off")
plt.show()
```
