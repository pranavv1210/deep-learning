# LSTM and GRU

A beginner-friendly introduction to lstm and gru.


## Why LSTM and GRU?

LSTM and GRU are gated recurrent architectures designed to preserve useful information over longer sequences than a simple RNN often can. Gates control what information is kept, updated or exposed.

## Learning Goals

- Understand why gated RNNs exist
- Compare LSTM and GRU concepts
- Build a tiny sequence model


### LSTM gates

An LSTM uses a cell state plus gates commonly described as forget, input and output gates. The forget gate controls what old information to retain; the input gate controls what new information to add; the output gate controls what becomes the hidden output.

## Concept Notes

LSTM and GRU layers use gates to decide what to remember and forget. This helps with longer sequences.


### GRU gates

A GRU uses update and reset gates and combines some LSTM mechanisms into a simpler structure. It has fewer components and parameters, but neither architecture is universally better; compare them on the task.

## Step-by-Step Example

Run this compact example first. Focus on the inputs, the transformation, and the output.


## Key terms

- **Gate:** learned control value, often between 0 and 1.
- **Cell state:** memory pathway in an LSTM.
- **Forget gate:** controls retention of earlier state.
- **Update gate:** controls how much new state replaces old state in a GRU.
- **Reset gate:** controls how much previous information contributes to candidate state.

```python
try:
    import tensorflow as tf
    model = tf.keras.Sequential([
        tf.keras.layers.Input(shape=(5, 1)),
        tf.keras.layers.GRU(8),
        tf.keras.layers.Dense(1)
    ])
    model.compile(optimizer="adam", loss="mse")
    model.summary()
except Exception as error:
    print("TensorFlow is not installed. Install requirements.txt to run this notebook fully.")
    print("Environment detail:", error)
```
