## Variables and Loops

Variables store values. Loops repeat a step, which is useful for training epochs and data processing.


```python
learning_rate = 0.01
epochs = 5
losses = []

for epoch in range(1, epochs + 1):
    loss = 1 / epoch
    losses.append(loss)
    print(f"Epoch {epoch}: loss={loss:.3f}")

print("Loss history:", losses)
```

## Functions and Classes

Functions keep repeated logic tidy. Classes are often used for datasets and models.


```python
def predict_linear(x, weight, bias):
    return weight * x + bias

class TinyModel:
    def __init__(self, weight, bias):
        self.weight = weight
        self.bias = bias

    def predict(self, x):
        return predict_linear(x, self.weight, self.bias)

model = TinyModel(weight=2.0, bias=1.0)
print(model.predict(5))
```

## Basic File Handling

Experiments often save metrics, logs, or model settings to files.


```python
metrics_path = "training_log.txt"

with open(metrics_path, "w", encoding="utf-8") as file:
    file.write("epoch,loss\n")
    for epoch, loss in enumerate(losses, start=1):
        file.write(f"{epoch},{loss:.3f}\n")

with open(metrics_path, "r", encoding="utf-8") as file:
    print(file.read())
```
