# Data Augmentation

A beginner-friendly introduction to data augmentation.


## Learning Goals

- Explain why augmentation helps
- Apply flips, noise, and rotations
- Visualize augmented samples


## Concept Notes

Data augmentation creates realistic variations of training examples so a model learns more robust patterns.


## Step-by-Step Example

Run this compact example first. Focus on the inputs, the transformation, and the output.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.ndimage import rotate

image = np.zeros((32, 32))
image[10:22, 14:18] = 1
augmented = [image, np.fliplr(image), rotate(image, 20, reshape=False), image + np.random.normal(0, 0.05, image.shape)]

fig, axes = plt.subplots(1, 4, figsize=(10, 3))
for ax, img, title in zip(axes, augmented, ["original", "flip", "rotate", "noise"]):
    ax.imshow(img, cmap="gray")
    ax.set_title(title)
    ax.axis("off")
plt.show()
```
