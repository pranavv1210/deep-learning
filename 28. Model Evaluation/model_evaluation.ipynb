# Model Evaluation

A beginner-friendly introduction to model evaluation.


## Learning Goals

- Compute accuracy, precision, recall, and F1
- Read a confusion matrix
- Use ROC-AUC for probabilities


## Concept Notes

Evaluation checks how well a model performs beyond training loss. Classification metrics highlight different kinds of errors.


## Step-by-Step Example

Run this compact example first. Focus on the inputs, the transformation, and the output.


```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, roc_auc_score

y_true = [1, 0, 1, 1, 0, 0, 1]
y_pred = [1, 0, 1, 0, 0, 1, 1]
y_prob = [0.9, 0.2, 0.8, 0.4, 0.1, 0.7, 0.6]

print("accuracy:", accuracy_score(y_true, y_pred))
print("precision:", precision_score(y_true, y_pred))
print("recall:", recall_score(y_true, y_pred))
print("f1:", f1_score(y_true, y_pred))
print("confusion matrix:\n", confusion_matrix(y_true, y_pred))
print("roc auc:", roc_auc_score(y_true, y_prob))
```
