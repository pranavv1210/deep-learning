## Load a Small Dataset

This notebook uses a tiny local CSV so it works offline.


```python
import pandas as pd

df = pd.read_csv("data/students.csv")
df.head()
```

## Filtering and New Columns

You can filter rows using conditions and create features from existing columns.


```python
df["passed"] = df["score"] >= 70
high_scores = df[df["score"] >= 80]

print(high_scores)
print(df[["name", "score", "passed"]])
```

## Missing Values and GroupBy

Clean missing values before modeling. GroupBy helps compare categories.


```python
df["hours_studied"] = df["hours_studied"].fillna(df["hours_studied"].median())
summary = df.groupby("course")["score"].agg(["mean", "min", "max"])
summary
```
