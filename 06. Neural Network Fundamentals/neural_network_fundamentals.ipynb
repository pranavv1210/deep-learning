## Neural networks in plain language

A neural network is a function made from connected layers of simple calculations. It receives numeric inputs, combines them using learned weights and biases, applies non-linear activation functions, and produces an output. Training changes the weights so predictions become more accurate on examples.

## Learning Goals

- Explain neurons, weights, bias, and layers
- Compute a forward pass by hand
- Visualize a simple network output


### Neurons, weights and bias

A neuron first calculates a weighted sum: `z = w1*x1 + w2*x2 + ... + b`. The inputs are x values, weights decide how strongly each input contributes, and bias shifts the result. An activation function then transforms z into the neuron's output.

## Concept Notes

A neural network is a composition of layers. Each neuron calculates a weighted sum, adds a bias, and usually passes the result through an activation function.


### Layers and forward pass

An input layer receives features. Hidden layers transform them into increasingly useful representations. The output layer produces the prediction, such as a class score or numeric value. Computing from input to output is the **forward pass**.

## Step-by-Step Example

Run this compact example first. Focus on the inputs, the transformation, and the output.


### Why non-linearity matters

If every layer only performs a linear transformation, stacking layers is still equivalent to one linear transformation. Activation functions introduce non-linearity so a network can represent more complex patterns.

```python
import numpy as np

inputs = np.array([0.5, -1.0, 2.0])
weights = np.array([0.4, -0.7, 0.2])
bias = 0.1

z = np.dot(inputs, weights) + bias
activation = max(0, z)
print("Weighted sum:", z)
print("ReLU output:", activation)
```

## Key terms

- **Neuron/unit:** one weighted calculation followed by an activation.
- **Weight:** trainable multiplier on an input.
- **Bias:** trainable offset.
- **Layer:** group of neurons operating at the same stage.
- **Hidden layer:** layer between input and output.
- **Forward pass:** compute the network prediction.
- **Activation:** transformed output of a neuron.

A network is not automatically intelligent because it has many layers. Its performance depends on data quality, architecture, training objective, optimization and evaluation.
