# Word Embeddings

A beginner-friendly introduction to word embeddings.


## Learning Goals

- Represent words as dense vectors
- Compare one-hot and embedding ideas
- Use simple similarity


## Concept Notes

Embeddings map words to dense numeric vectors. Similar meanings can be placed near each other in vector space.


## Step-by-Step Example

Run this compact example first. Focus on the inputs, the transformation, and the output.


```python
import numpy as np

embeddings = {
    "king": np.array([0.8, 0.6, 0.1]),
    "queen": np.array([0.78, 0.62, 0.12]),
    "car": np.array([0.1, 0.2, 0.9]),
}

def cosine_similarity(a, b):
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))

print("king vs queen:", round(cosine_similarity(embeddings["king"], embeddings["queen"]), 3))
print("king vs car:", round(cosine_similarity(embeddings["king"], embeddings["car"]), 3))
```
