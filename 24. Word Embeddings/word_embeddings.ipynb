# Word Embeddings

A beginner-friendly introduction to word embeddings.


## Word embeddings

A word embedding maps a token to a dense numeric vector. Similar words can have vectors with related directions or nearby positions, allowing models to use semantic and syntactic patterns rather than treating every token as an unrelated ID.

## Learning Goals

- Represent words as dense vectors
- Compare one-hot and embedding ideas
- Use simple similarity


### Sparse counts versus dense vectors

A one-hot vector has one position per vocabulary item and only one active value, so it is large and does not express similarity. An embedding uses a shorter learned vector with many continuous values. The vector dimensions do not usually have simple human-readable meanings.

## Concept Notes

Embeddings map words to dense numeric vectors. Similar meanings can be placed near each other in vector space.


### Context and limitations

Static embeddings such as Word2Vec assign one vector per word, so a word with multiple meanings has one representation. Contextual models create representations based on surrounding tokens. Similarity in embedding space is useful but can also encode biases present in training data.

## Step-by-Step Example

Run this compact example first. Focus on the inputs, the transformation, and the output.


## Key terms

- **Embedding:** learned dense vector representation.
- **One-hot encoding:** vector with one active position for a category.
- **Vocabulary:** set of known tokens.
- **Contextual embedding:** representation that depends on surrounding tokens.
- **Cosine similarity:** normalized dot product measuring vector direction similarity.

```python
import numpy as np

embeddings = {
    "king": np.array([0.8, 0.6, 0.1]),
    "queen": np.array([0.78, 0.62, 0.12]),
    "car": np.array([0.1, 0.2, 0.9]),
}

def cosine_similarity(a, b):
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))

print("king vs queen:", round(cosine_similarity(embeddings["king"], embeddings["queen"]), 3))
print("king vs car:", round(cosine_similarity(embeddings["king"], embeddings["car"]), 3))
```
