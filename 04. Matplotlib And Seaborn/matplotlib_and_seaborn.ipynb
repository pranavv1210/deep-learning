# Matplotlib and Seaborn

Create beginner-friendly visualizations for data exploration.


## Learning Goals

- Make line, bar, histogram, scatter, and heatmap plots
- Label charts clearly
- Use plots to understand distributions and relationships


## Concept Notes

Visualization helps you notice trends, outliers, and class imbalance before training a model.


## Line and Bar Plots

Start with simple charts and always label axes.


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("data/training_metrics.csv")

plt.plot(df["epoch"], df["loss"], marker="o", label="loss")
plt.plot(df["epoch"], df["accuracy"], marker="o", label="accuracy")
plt.xlabel("Epoch")
plt.title("Training Metrics")
plt.legend()
plt.show()
```

## Histogram and Scatter Plot

Histograms show distribution. Scatter plots show relationships between two variables.


```python
students = pd.read_csv("../03. Pandas/data/students.csv")

sns.histplot(students["score"], bins=5, kde=True)
plt.title("Score Distribution")
plt.show()

sns.scatterplot(data=students, x="hours_studied", y="score", hue="course")
plt.title("Study Time vs Score")
plt.show()
```

## Heatmap

A heatmap is useful for matrices such as correlations or confusion matrices.


```python
numeric = students[["hours_studied", "score"]].corr()
sns.heatmap(numeric, annot=True, cmap="Blues", vmin=-1, vmax=1)
plt.title("Correlation Heatmap")
plt.show()
```
