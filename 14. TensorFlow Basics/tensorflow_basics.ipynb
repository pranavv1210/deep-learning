# TensorFlow Basics

A beginner-friendly introduction to tensorflow basics.


## TensorFlow overview

TensorFlow is a framework for numerical computation and machine learning. It provides tensors, automatic differentiation, model-building tools, data pipelines and hardware acceleration. A TensorFlow tensor is similar to a NumPy array but can participate in TensorFlow's computation and gradient systems.

## Learning Goals

- Create TensorFlow tensors
- Run tensor operations
- Build a tiny model


### Eager execution and graphs

In eager execution, operations run immediately and return results. TensorFlow can also trace functions into graphs for optimization and deployment. A `tf.function`-decorated function may behave differently from ordinary Python because TensorFlow traces supported operations.

## Concept Notes

TensorFlow provides tensors, automatic differentiation, and high-level model-building tools.


### Gradients

`tf.GradientTape` records operations involving watched tensors. Calling `gradient` calculates derivatives of a target value with respect to selected variables. This is how custom training loops obtain gradients.

## Step-by-Step Example

Run this compact example first. Focus on the inputs, the transformation, and the output.


## Key terms

- **Tensor:** multi-dimensional numeric data.
- **Automatic differentiation:** computation of derivatives by applying derivative rules to recorded operations.
- **Eager execution:** operations execute immediately.
- **Graph tracing:** record computations into an optimized graph.
- **GradientTape:** TensorFlow context that records operations for gradient calculation.

```python
try:
    import tensorflow as tf
    x = tf.constant([[1.0, 2.0], [3.0, 4.0]])
    print("Tensor:", x)
    print("Mean:", tf.reduce_mean(x).numpy())
except Exception as error:
    print("TensorFlow is not installed. Install requirements.txt to run this notebook fully.")
    print("Environment detail:", error)
```
