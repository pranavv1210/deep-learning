## Concept Notes

PyTorch is a flexible deep learning framework. Models are Python classes that define a forward pass.


## Step-by-Step Example

Run this compact example first. Focus on the inputs, the transformation, and the output.


```python
try:
    import torch
    from torch import nn
    from torch.utils.data import TensorDataset, DataLoader

    X = torch.randn(20, 3)
    y = (X.sum(dim=1) > 0).float().unsqueeze(1)
    loader = DataLoader(TensorDataset(X, y), batch_size=5, shuffle=True)

    class TinyNet(nn.Module):
        def __init__(self):
            super().__init__()
            self.layers = nn.Sequential(nn.Linear(3, 8), nn.ReLU(), nn.Linear(8, 1), nn.Sigmoid())

        def forward(self, x):
            return self.layers(x)

    model = TinyNet()
    for batch_X, batch_y in loader:
        print(model(batch_X).shape)
        break
except ImportError:
    print("PyTorch is not installed. Install requirements.txt to run this notebook fully.")
```
