## Load, Train, Evaluate, Predict

This cell keeps the full starter pipeline together so you can run it end to end.


```python
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report

X, y = make_classification(n_samples=500, n_features=20, n_informative=8, weights=[0.7, 0.3], random_state=42)
X_train, X_test, y_train, y_test = train_test_split(X, y, stratify=y, random_state=42)
model = LogisticRegression(max_iter=1000, random_state=42).fit(X_train, y_train)
print(classification_report(y_test, model.predict(X_test)))
```
