## Goal

Text classification assigns a predefined category to a document or message, such as topic, spam status or support-ticket type.

# Text Classification

Classify short text into topic categories.


## Features and evaluation

A bag-of-words or TF-IDF representation converts tokens into numeric features. TF-IDF gives less weight to terms common across many documents and more weight to terms that are distinctive to a document. Split before fitting the vectorizer to prevent vocabulary leakage.

## Learning Goals

- Define the task
- Train a simple baseline
- Evaluate and make predictions


## Key terms

- **TF-IDF:** term frequency times inverse document frequency.
- **Sparse matrix:** matrix in which most values are zero, stored efficiently.
- **N-gram:** sequence of n adjacent tokens.
- **Vocabulary leakage:** test documents influence which features are learned during training.

## Load, Train, Evaluate, Predict

This cell keeps the full starter pipeline together so you can run it end to end.


```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import Pipeline

texts = ['team wins match', 'player scores goal', 'new AI model', 'python neural network', 'stock market rises', 'investors buy shares']
labels = ['sports', 'sports', 'tech', 'tech', 'finance', 'finance']
model = Pipeline([('tfidf', TfidfVectorizer()), ('classifier', MultinomialNB())])
model.fit(texts, labels)
print(model.predict(['AI scores new benchmark', 'market shares rise']))
```
