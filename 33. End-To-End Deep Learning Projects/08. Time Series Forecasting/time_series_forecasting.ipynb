## Load, Train, Evaluate, Predict

This cell keeps the full starter pipeline together so you can run it end to end.


```python
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error

values = np.sin(np.arange(120) / 6) + np.random.normal(0, 0.05, 120)
X, y = [], []
for i in range(3, len(values)):
    X.append(values[i-3:i])
    y.append(values[i])
X, y = np.array(X), np.array(y)
model = LinearRegression().fit(X[:90], y[:90])
predictions = model.predict(X[90:])
print('MAE:', mean_absolute_error(y[90:], predictions))
print('Next prediction:', predictions[-1])
```
