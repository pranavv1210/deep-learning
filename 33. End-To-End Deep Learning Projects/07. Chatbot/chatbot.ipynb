## Load, Train, Evaluate, Predict

This cell keeps the full starter pipeline together so you can run it end to end.


```python
responses = {
    'hello': 'Hi! What deep learning topic are you studying?',
    'cnn': 'CNNs are useful for images because they learn local patterns.',
    'rnn': 'RNNs process sequences one step at a time.',
}

def reply(message):
    message = message.lower()
    for keyword, answer in responses.items():
        if keyword in message:
            return answer
    return 'I do not know that yet. Add a new rule and try again.'

print(reply('hello'))
print(reply('tell me about cnn'))
```
