## Load, Train, Evaluate, Predict

This cell keeps the full starter pipeline together so you can run it end to end.


```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline

texts = ['great movie', 'loved this', 'excellent work', 'bad movie', 'terrible plot', 'not good']
labels = [1, 1, 1, 0, 0, 0]
model = Pipeline([('vectorizer', CountVectorizer()), ('classifier', LogisticRegression())])
model.fit(texts, labels)
print(model.predict(['excellent movie', 'terrible work']))
```
