# Project: Object Detection

## Goal

Object detection identifies object categories and their locations in an image. Unlike image classification, it can return multiple objects with bounding boxes.

# Object Detection

Predict a toy bounding box around a bright square in synthetic images.


## Bounding boxes and predictions

A bounding box is commonly represented by coordinates such as x-min, y-min, x-max, y-max or center-x, center-y, width and height. A detector predicts boxes, class scores and confidence values. Coordinate conventions and normalization rules must match the model.

## Learning Goals

- Define the task
- Train a simple baseline
- Evaluate and make predictions


## Evaluate detection

Intersection over Union (IoU) measures overlap between a predicted box and a ground-truth box. Precision and recall are evaluated across confidence thresholds; mean Average Precision (mAP) summarizes detection performance across classes and thresholds under a specified protocol.

## Key terms

- **Bounding box:** rectangle around an object.
- **IoU:** overlap divided by union area of two boxes.
- **Confidence score:** model's confidence in a detection.
- **Non-maximum suppression:** remove duplicate overlapping detections.
- **mAP:** mean Average Precision over categories under a chosen evaluation protocol.

## Load, Train, Evaluate, Predict

This cell keeps the full starter pipeline together so you can run it end to end.


```python
import numpy as np
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split

images, boxes = [], []
for _ in range(250):
    img = np.zeros((16, 16))
    x, y = np.random.randint(0, 11), np.random.randint(0, 11)
    img[y:y+5, x:x+5] = 1
    images.append(img.ravel())
    boxes.append([x, y, x + 5, y + 5])
X_train, X_test, y_train, y_test = train_test_split(np.array(images), np.array(boxes), random_state=42)
model = RandomForestRegressor(random_state=42).fit(X_train, y_train)
print('True box:', y_test[0])
print('Predicted box:', np.round(model.predict(X_test[:1])[0], 1))
```
