## Load, Train, Evaluate, Predict

This cell keeps the full starter pipeline together so you can run it end to end.


```python
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

digits = load_digits()
X_train, X_test, y_train, y_test = train_test_split(digits.data / 16.0, digits.target, random_state=42)
model = KNeighborsClassifier(n_neighbors=3).fit(X_train, y_train)
print('Accuracy:', accuracy_score(y_test, model.predict(X_test)))
print('First 10 predictions:', model.predict(X_test[:10]))
```
