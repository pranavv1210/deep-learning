## Concept Notes

End-to-end projects connect the ideas from earlier notebooks. The goal is not to build the largest model first. The goal is to build a complete, understandable pipeline and then improve it step by step.


## Project Workflow

Every project in this section follows the same beginner-friendly pipeline.


```python
steps = [
    "define the problem",
    "load the dataset",
    "preprocess the data",
    "build a baseline model",
    "train and evaluate",
    "show predictions",
    "save or improve the model",
]

for number, step in enumerate(steps, start=1):
    print(f"{number}. {step}")
```

## Choose a Baseline

A baseline gives you a simple first result. Improve only after the baseline works.


```python
project_baselines = {
    "image classification": "MLP or small CNN",
    "sentiment analysis": "TF-IDF + logistic regression",
    "time series": "lag features + linear regression",
    "recommendation": "user-item similarity",
}

for project, baseline in project_baselines.items():
    print(f"{project}: {baseline}")
```
