# Autoencoders

A beginner-friendly introduction to autoencoders.


## Autoencoders

An autoencoder learns to reconstruct its input. The encoder compresses the input into a latent representation; the decoder uses that representation to rebuild the input. Training minimizes reconstruction error.

## Learning Goals

- Explain encoder-decoder structure
- Compress and reconstruct data
- Measure reconstruction error


### Latent space and bottleneck

The latent vector is a compact internal representation. A bottleneck forces the model to preserve the most useful information in a smaller space. The latent representation is not automatically meaningful or disentangled; that depends on model design and training.

## Concept Notes

An autoencoder learns to copy input through a small bottleneck. The bottleneck encourages compact representations.


### Applications and limitations

Autoencoders can support dimensionality reduction, denoising and anomaly detection. For anomaly detection, reconstruction error may be higher for unusual examples, but thresholds need validation and anomalous data must be represented appropriately.

## Step-by-Step Example

Run this compact example first. Focus on the inputs, the transformation, and the output.


## Key terms

- **Encoder:** maps input to latent representation.
- **Decoder:** reconstructs input from latent representation.
- **Latent space:** compressed representation space.
- **Reconstruction loss:** error between input and reconstructed output.
- **Bottleneck:** constrained intermediate representation.

```python
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.neural_network import MLPRegressor
from sklearn.metrics import mean_squared_error

digits = load_digits()
X = digits.data / 16.0
X_train, X_test = train_test_split(X, random_state=42)

autoencoder = MLPRegressor(hidden_layer_sizes=(16, 8, 16), max_iter=300, random_state=42)
autoencoder.fit(X_train, X_train)
reconstructed = autoencoder.predict(X_test)
print("Reconstruction MSE:", mean_squared_error(X_test, reconstructed))
```
