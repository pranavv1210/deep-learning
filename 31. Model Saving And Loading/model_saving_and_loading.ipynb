# Model Saving and Loading

A beginner-friendly introduction to model saving and loading.


## Save a trained neural network

Model saving stores learned parameters and, depending on the format, model architecture and training state. Saving lets you resume training or load the model for inference without starting from scratch.

## Learning Goals

- Save Python objects with pickle
- Save scikit-learn models with joblib
- Understand TensorFlow and PyTorch saving


### Weights versus full model

Saving weights alone means you must recreate the same architecture before loading them. Saving a full model format can include architecture and sometimes optimizer state. Framework formats and compatibility rules differ, so follow the matching official documentation.

## Concept Notes

Saving a model lets you reuse it for inference without retraining every time.


### Reproducibility and security

Record framework version, preprocessing, input shape, label mapping and training configuration. Do not load model files from untrusted sources without understanding the format and security implications.

## Step-by-Step Example

Run this compact example first. Focus on the inputs, the transformation, and the output.


## Key terms

- **Checkpoint:** saved model state at a particular training point.
- **Weights:** learned numeric parameters.
- **Optimizer state:** information such as momentum buffers needed to resume training faithfully.
- **Inference artifact:** saved model package used for predictions.

```python
import joblib
from sklearn.datasets import make_classification
from sklearn.linear_model import LogisticRegression

X, y = make_classification(n_samples=200, n_features=4, random_state=42)
model = LogisticRegression(random_state=42)
model.fit(X, y)

joblib.dump(model, "logistic_model.joblib")
loaded_model = joblib.load("logistic_model.joblib")
print("Prediction:", loaded_model.predict(X[:3]))
```
