# Model Saving and Loading

A beginner-friendly introduction to model saving and loading.


## Learning Goals

- Save Python objects with pickle
- Save scikit-learn models with joblib
- Understand TensorFlow and PyTorch saving


## Concept Notes

Saving a model lets you reuse it for inference without retraining every time.


## Step-by-Step Example

Run this compact example first. Focus on the inputs, the transformation, and the output.


```python
import joblib
from sklearn.datasets import make_classification
from sklearn.linear_model import LogisticRegression

X, y = make_classification(n_samples=200, n_features=4, random_state=42)
model = LogisticRegression(random_state=42)
model.fit(X, y)

joblib.dump(model, "logistic_model.joblib")
loaded_model = joblib.load("logistic_model.joblib")
print("Prediction:", loaded_model.predict(X[:3]))
```
