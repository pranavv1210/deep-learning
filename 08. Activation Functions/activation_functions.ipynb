# Activation Functions

A beginner-friendly introduction to activation functions.


## Learning Goals

- Compare sigmoid, tanh, ReLU, and softmax
- Plot activation curves
- Understand where activations are used


## Concept Notes

Activation functions add non-linearity. Without them, stacked layers would still behave like one linear transformation.


## Step-by-Step Example

Run this compact example first. Focus on the inputs, the transformation, and the output.


```python
import numpy as np
import matplotlib.pyplot as plt

def sigmoid(x): return 1 / (1 + np.exp(-x))
def relu(x): return np.maximum(0, x)
def softmax(x):
    shifted = x - np.max(x)
    exp_values = np.exp(shifted)
    return exp_values / exp_values.sum()

x = np.linspace(-5, 5, 200)
plt.plot(x, sigmoid(x), label="sigmoid")
plt.plot(x, np.tanh(x), label="tanh")
plt.plot(x, relu(x), label="relu")
plt.legend()
plt.title("Activation Functions")
plt.show()

print("softmax:", softmax(np.array([1.0, 2.0, 3.0])))
```
