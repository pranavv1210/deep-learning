# Activation Functions

A beginner-friendly introduction to activation functions.


## What is an activation function?

An activation function transforms a neuron's weighted sum into its output. Without non-linear activations, a deep stack of linear layers would still be equivalent to one linear transformation. Activations let networks represent non-linear patterns.

## Learning Goals

- Compare sigmoid, tanh, ReLU, and softmax
- Plot activation curves
- Understand where activations are used


### Common activation functions

- **Sigmoid:** maps values to (0, 1); historically common in binary outputs, but can saturate in hidden layers.
- **Tanh:** maps values to (-1, 1); centered around zero but can also saturate.
- **ReLU:** `max(0, x)`; simple and common in hidden layers, but units can become inactive for negative inputs.
- **Leaky ReLU:** keeps a small slope for negative inputs.
- **Softmax:** converts a vector of class scores into probabilities that sum to one.

## Concept Notes

Activation functions add non-linearity. Without them, stacked layers would still behave like one linear transformation.


### Saturation and gradients

A function **saturates** when large positive or negative inputs produce outputs that change very little. Its derivative becomes small, so gradient-based learning can become slow through that activation. This is one reason ReLU-family activations are common in hidden layers.

## Step-by-Step Example

Run this compact example first. Focus on the inputs, the transformation, and the output.


## Key terms

- **Non-linearity:** behavior that cannot be represented by one linear function.
- **Saturation:** output changes very little as input changes.
- **Gradient:** rate of change used to update model parameters.
- **Dead ReLU:** a ReLU unit that remains on the zero-output side for its inputs and receives no useful gradient through that path.
- **Softmax:** normalization that produces a probability distribution over classes.

```python
import numpy as np
import matplotlib.pyplot as plt

def sigmoid(x): return 1 / (1 + np.exp(-x))
def relu(x): return np.maximum(0, x)
def softmax(x):
    shifted = x - np.max(x)
    exp_values = np.exp(shifted)
    return exp_values / exp_values.sum()

x = np.linspace(-5, 5, 200)
plt.plot(x, sigmoid(x), label="sigmoid")
plt.plot(x, np.tanh(x), label="tanh")
plt.plot(x, relu(x), label="relu")
plt.legend()
plt.title("Activation Functions")
plt.show()

print("softmax:", softmax(np.array([1.0, 2.0, 3.0])))
```
