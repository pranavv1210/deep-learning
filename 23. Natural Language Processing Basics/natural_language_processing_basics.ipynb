# Natural Language Processing Basics

A beginner-friendly introduction to natural language processing basics.


## What is NLP?

Natural Language Processing (NLP) is the area of AI that works with human language. Text is unstructured from a computer's perspective, so NLP systems convert it into numeric representations that algorithms can process.

## Learning Goals

- Tokenize text
- Build a vocabulary
- Pad sequences for batching


### Text preprocessing

Tokenization splits text into units such as words or subwords. Lowercasing can reduce vocabulary size but may erase meaning in some tasks. Stop-word removal, stemming and lemmatization are optional techniques, not mandatory steps: they can remove useful context, especially for modern transformer models.

## Concept Notes

NLP turns text into numbers. The common early steps are tokenization, vocabulary building, integer encoding, and padding.


### Common NLP tasks

- Text classification: assign a label to text, such as spam or topic.
- Sentiment analysis: estimate opinion or emotional polarity.
- Named entity recognition: identify names, organizations, places and other entities.
- Machine translation: convert text between languages.
- Question answering and summarization: produce answers or shorter representations.

## Step-by-Step Example

Run this compact example first. Focus on the inputs, the transformation, and the output.


## Key terms

- **Corpus:** collection of text documents.
- **Token:** unit produced by tokenization.
- **Vocabulary:** set of tokens known to a text model.
- **Normalization:** consistent text transformation.
- **Stemming:** heuristic reduction of words to stems.
- **Lemmatization:** mapping words to dictionary base forms.
- **N-gram:** sequence of n adjacent tokens.

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

sentences = ["deep learning is useful", "learning text models", "deep models learn patterns"]
tokens = [sentence.lower().split() for sentence in sentences]
vocab = {word: index + 1 for index, word in enumerate(sorted({word for row in tokens for word in row}))}
encoded = [[vocab[word] for word in row] for row in tokens]
max_length = max(len(row) for row in encoded)
padded = [row + [0] * (max_length - len(row)) for row in encoded]

print("vocab:", vocab)
print("padded:", padded)
```
