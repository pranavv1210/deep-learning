# Optimizers

A beginner-friendly introduction to optimizers.


## What does an optimizer do?

An optimizer updates trainable weights and biases using gradients calculated from the loss. Gradient descent is the basic idea: move parameters in the direction that reduces loss. Optimizers differ in how they scale, smooth and combine gradient updates.

## Learning Goals

- Understand SGD, momentum, and Adam at a high level
- Compare update rules
- See how learning rate changes progress


### SGD and momentum

Stochastic Gradient Descent (SGD) updates parameters using a mini-batch rather than the entire dataset. Momentum keeps part of earlier updates, which can reduce zig-zagging and accelerate movement in consistent directions.

## Concept Notes

Optimizers update model parameters using gradients. The learning rate controls the size of each update.


### Adaptive optimizers

Adam combines momentum-like first-moment estimates with adaptive scaling based on squared gradients. RMSprop also adapts updates using a moving average of squared gradients. Adaptive methods are convenient defaults, but they are not guaranteed to outperform SGD in every task.

## Step-by-Step Example

Run this compact example first. Focus on the inputs, the transformation, and the output.


### Learning rate

The learning rate sets the size of updates. If it is too large, training can oscillate or diverge; if too small, progress may be extremely slow. Learning-rate schedules reduce or change the rate during training.

```python
import numpy as np
import matplotlib.pyplot as plt

def gradient(x):
    return 2 * (x - 3)

x = 8.0
learning_rate = 0.2
history = []
for step in range(20):
    x -= learning_rate * gradient(x)
    history.append(x)

plt.plot(history, marker="o")
plt.axhline(3, color="black", linestyle="--", label="minimum")
plt.title("SGD Moving Toward Minimum")
plt.legend()
plt.show()
print("Final x:", round(x, 4))
```

## Key terms

- **Optimizer:** algorithm updating trainable parameters.
- **Mini-batch:** subset of examples used for one gradient update.
- **Momentum:** moving direction based partly on previous gradients.
- **Adaptive learning rate:** per-parameter update scaling based on gradient history.
- **Learning-rate schedule:** rule changing learning rate during training.
