# Image Data Preprocessing

A beginner-friendly introduction to image data preprocessing.


## Learning Goals

- Represent images as arrays
- Resize and normalize image data
- Apply simple augmentation ideas


## Concept Notes

Images are tensors with height, width, and channels. Preprocessing usually resizes, normalizes, and sometimes augments images.


## Step-by-Step Example

Run this compact example first. Focus on the inputs, the transformation, and the output.


```python
import numpy as np
import matplotlib.pyplot as plt
from skimage.transform import resize

image = np.zeros((32, 32, 3))
image[8:24, 10:22, 0] = 1.0
resized = resize(image, (16, 16), anti_aliasing=True)
flipped = np.fliplr(resized)

fig, axes = plt.subplots(1, 3, figsize=(8, 3))
for ax, img, title in zip(axes, [image, resized, flipped], ["Original", "Resized", "Flipped"]):
    ax.imshow(img)
    ax.set_title(title)
    ax.axis("off")
plt.show()
```
