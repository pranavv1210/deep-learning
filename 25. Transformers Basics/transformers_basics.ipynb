# Transformers Basics

A beginner-friendly introduction to transformers basics.


## Learning Goals

- Understand attention
- Compute simple attention weights
- Connect attention to transformer blocks


## Concept Notes

Transformers use attention so each token can look at other tokens and decide what information matters.


## Step-by-Step Example

Run this compact example first. Focus on the inputs, the transformation, and the output.


```python
import numpy as np

def softmax(x):
    exp_values = np.exp(x - np.max(x))
    return exp_values / exp_values.sum()

query = np.array([1.0, 0.0])
keys = np.array([[1.0, 0.0], [0.5, 0.5], [0.0, 1.0]])
values = np.array([[10, 0], [5, 5], [0, 10]])

scores = keys @ query
weights = softmax(scores)
context = weights @ values

print("attention weights:", np.round(weights, 3))
print("context vector:", np.round(context, 3))
```
