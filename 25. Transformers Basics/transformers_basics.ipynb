# Transformers Basics

A beginner-friendly introduction to transformers basics.


## Why transformers?

Transformers process token sequences using attention rather than recurrence. Attention lets each token combine information from other tokens, which makes it easier to model long-range relationships and parallelize much of training.

## Learning Goals

- Understand attention
- Compute simple attention weights
- Connect attention to transformer blocks


### Queries, keys and values

For each token, the model computes a query, key and value vector. Similarity between a query and keys produces attention scores. After scaling and softmax normalization, those scores weight the value vectors to create a context-aware representation.

## Concept Notes

Transformers use attention so each token can look at other tokens and decide what information matters.


### Scaled dot-product attention

The core formula is `Attention(Q, K, V) = softmax(QK^T / sqrt(d_k)) V`. Q contains queries, K keys and V values; `d_k` is key-vector dimension. Dividing by `sqrt(d_k)` helps keep dot-product magnitudes in a range that makes softmax gradients more stable.

## Step-by-Step Example

Run this compact example first. Focus on the inputs, the transformation, and the output.


### Multi-head attention and positional information

Multi-head attention performs several attention operations in parallel so the model can learn different relationship patterns. Since attention alone does not encode token order, transformers need positional information such as positional embeddings or encodings.

```python
import numpy as np

def softmax(x):
    exp_values = np.exp(x - np.max(x))
    return exp_values / exp_values.sum()

query = np.array([1.0, 0.0])
keys = np.array([[1.0, 0.0], [0.5, 0.5], [0.0, 1.0]])
values = np.array([[10, 0], [5, 5], [0, 10]])

scores = keys @ query
weights = softmax(scores)
context = weights @ values

print("attention weights:", np.round(weights, 3))
print("context vector:", np.round(context, 3))
```

## Key terms

- **Attention:** weighted combination of information from tokens.
- **Query:** representation describing what information a token seeks.
- **Key:** representation used to match candidate information.
- **Value:** content combined using attention weights.
- **Head:** one learned attention operation.
- **Positional encoding/embedding:** information indicating token order.
- **Context window:** maximum sequence context considered by a model.
