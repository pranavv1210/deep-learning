# Artificial Neural Networks

A beginner-friendly introduction to artificial neural networks.


## Artificial Neural Networks (ANNs)

An ANN is a network of connected artificial neurons. Dense or fully connected layers connect each unit in one layer to every unit in the next. Networks learn useful combinations of input features by updating weights and biases during training.

## Learning Goals

- Build a small feed-forward network
- Train on a simple classification dataset
- Evaluate model accuracy


### Forward pass, loss and backward pass

The forward pass produces predictions. A loss function compares predictions with targets. Backpropagation calculates gradients, and an optimizer updates the parameters. These steps repeat over many batches and epochs.

## Concept Notes

An artificial neural network connects dense layers to learn patterns in tabular or flattened data.


### Output layer depends on task

For regression, the output is often a linear numeric value. For binary classification, a sigmoid probability is common. For mutually exclusive multi-class classification, softmax is common. The loss function and label encoding must match the output format.

## Step-by-Step Example

Run this compact example first. Focus on the inputs, the transformation, and the output.


## Key terms

- **Dense layer:** fully connected layer.
- **Parameter:** learned weight or bias.
- **Epoch:** one pass through the training dataset.
- **Batch size:** number of examples used in one update.
- **Inference:** using the trained network to predict new examples.

```python
from sklearn.datasets import make_moons
from sklearn.model_selection import train_test_split
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score

X, y = make_moons(n_samples=300, noise=0.2, random_state=42)
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42)

model = MLPClassifier(hidden_layer_sizes=(16, 8), max_iter=800, random_state=42)
model.fit(X_train, y_train)
predictions = model.predict(X_test)
print("Accuracy:", accuracy_score(y_test, predictions))
```
