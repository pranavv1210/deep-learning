## Concept Notes

A GAN trains two models together: a generator creates fake samples, and a discriminator tries to distinguish real from fake.


## Step-by-Step Example

Run this compact example first. Focus on the inputs, the transformation, and the output.


```python
import numpy as np
import matplotlib.pyplot as plt

real_samples = np.random.normal(loc=4, scale=1, size=200)
fake_samples = np.random.normal(loc=0, scale=1, size=200)

plt.hist(real_samples, alpha=0.6, label="real")
plt.hist(fake_samples, alpha=0.6, label="fake")
plt.legend()
plt.title("GAN Intuition: Match the Real Distribution")
plt.show()

print("A trained generator would move the fake distribution toward the real one.")
```
