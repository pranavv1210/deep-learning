# Gradient Descent

A beginner-friendly introduction to gradient descent.


## Gradient descent intuition

Training a neural network means finding weights that make a loss function small. Imagine standing on a hill in fog: you can feel which direction slopes downward locally, take a step, and repeat. Gradient descent does this mathematically by calculating the gradient and moving parameters in the opposite direction.

## Learning Goals

- Compare batch, mini-batch, and stochastic updates
- Train a tiny linear model
- Plot loss over time


### Learning rate and convergence

The learning rate controls step size. If it is too small, convergence (settling toward a stable low-loss region) may be very slow. If it is too large, updates may jump across a minimum, oscillate or diverge. Loss curves help diagnose these behaviors.

## Concept Notes

Gradient descent reduces loss by stepping in the opposite direction of the gradient. Batch size changes how often updates happen.


### Batch, stochastic and mini-batch descent

Batch gradient descent calculates gradients using the entire dataset per update. Stochastic gradient descent uses one example. Mini-batch gradient descent uses a small group and is the common deep-learning choice because it balances noisy updates and computational efficiency.

## Step-by-Step Example

Run this compact example first. Focus on the inputs, the transformation, and the output.


## Key terms

- **Gradient:** vector of partial derivatives of loss.
- **Learning rate:** update step-size multiplier.
- **Convergence:** progress toward a stable solution.
- **Epoch:** one pass through all training examples.
- **Iteration/step:** one parameter update, often from one mini-batch.
- **Mini-batch:** group of examples used to calculate an update.

```python
import numpy as np
import matplotlib.pyplot as plt

X = np.linspace(0, 10, 50)
y = 2 * X + 1 + np.random.normal(0, 1, size=50)
w, b = 0.0, 0.0
lr = 0.01
losses = []

for epoch in range(100):
    y_pred = w * X + b
    error = y_pred - y
    w -= lr * np.mean(error * X)
    b -= lr * np.mean(error)
    losses.append(np.mean(error ** 2))

plt.plot(losses)
plt.title("Batch Gradient Descent Loss")
plt.show()
print("Learned weight and bias:", round(w, 2), round(b, 2))
```
