# Keras Basics

A beginner-friendly introduction to keras basics.


## Keras overview

Keras is a high-level deep-learning API for defining, training and evaluating neural networks. It helps organize layers and training configuration so you can focus on the model design instead of writing every low-level operation.

## Learning Goals

- Create a Sequential model
- Compile, fit, and evaluate
- Make predictions


### Sequential and Functional APIs

The Sequential API is appropriate for a simple stack where each layer has one input and one output. The Functional API supports branching, multiple inputs/outputs and shared layers. Choose the simplest API that represents the architecture.

## Concept Notes

Keras is TensorFlow's beginner-friendly model API. It lets you stack layers and train with a few clear steps.


### Compile, fit and evaluate

`compile` configures the optimizer, loss and metrics. `fit` trains the model on examples and can use validation data to monitor generalization. `evaluate` calculates configured metrics on supplied data. Do not use the test set repeatedly to make training decisions.

## Step-by-Step Example

Run this compact example first. Focus on the inputs, the transformation, and the output.


## Key terms

- **Layer:** reusable computation block in a network.
- **Sequential model:** linear stack of layers.
- **Functional API:** graph-style model definition supporting complex connections.
- **Compile:** configure training objective and optimization.
- **Fit:** train model parameters.
- **Validation data:** held-out examples used during development.

```python
try:
    import tensorflow as tf
    from sklearn.datasets import make_classification
    from sklearn.model_selection import train_test_split

    X, y = make_classification(n_samples=400, n_features=6, n_informative=4, random_state=42)
    X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42)

    model = tf.keras.Sequential([
        tf.keras.layers.Input(shape=(6,)),
        tf.keras.layers.Dense(12, activation="relu"),
        tf.keras.layers.Dense(1, activation="sigmoid"),
    ])
    model.compile(optimizer="adam", loss="binary_crossentropy", metrics=["accuracy"])
    model.fit(X_train, y_train, epochs=5, verbose=0)
    print(model.evaluate(X_test, y_test, verbose=0))
except Exception as error:
    print("TensorFlow is not installed. Install requirements.txt to run this notebook fully.")
    print("Environment detail:", error)
```
