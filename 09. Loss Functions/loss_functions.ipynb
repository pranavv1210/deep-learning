# Loss Functions

A beginner-friendly introduction to loss functions.


## Learning Goals

- Calculate MSE
- Calculate binary cross entropy
- Calculate categorical cross entropy


## Concept Notes

A loss function measures how wrong a prediction is. Training tries to reduce this number.


## Step-by-Step Example

Run this compact example first. Focus on the inputs, the transformation, and the output.


```python
import numpy as np

y_true = np.array([1, 0, 1, 1])
y_pred = np.array([0.9, 0.2, 0.4, 0.8])

mse = np.mean((y_true - y_pred) ** 2)
bce = -np.mean(y_true * np.log(y_pred) + (1 - y_true) * np.log(1 - y_pred))
print("MSE:", round(mse, 4))
print("Binary cross entropy:", round(bce, 4))

class_true = np.array([0, 1, 0])
class_pred = np.array([0.1, 0.8, 0.1])
cce = -np.sum(class_true * np.log(class_pred))
print("Categorical cross entropy:", round(cce, 4))
```
