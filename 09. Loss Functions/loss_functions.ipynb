# Loss Functions

A beginner-friendly introduction to loss functions.


## What is a loss function?

A loss function measures how wrong a prediction is compared with its target. Training tries to change the model parameters to reduce this loss. The correct loss depends on what the model predicts and how targets are represented.

## Learning Goals

- Calculate MSE
- Calculate binary cross entropy
- Calculate categorical cross entropy


### Regression losses

Mean squared error (MSE) averages squared numeric errors and strongly penalizes large mistakes. Mean absolute error (MAE) averages absolute errors and is less sensitive to very large deviations. Huber loss behaves quadratically for small errors and linearly for large errors.

## Concept Notes

A loss function measures how wrong a prediction is. Training tries to reduce this number.


### Classification losses

Binary cross-entropy is used for binary probability predictions. Categorical cross-entropy is used for multi-class probability outputs, often with one-hot targets. Sparse categorical cross-entropy uses integer class IDs instead. Match the output layer and target encoding to the loss.

## Step-by-Step Example

Run this compact example first. Focus on the inputs, the transformation, and the output.


## Key terms

- **Loss:** error value for one example or batch.
- **Objective:** quantity being optimized, sometimes loss plus regularization.
- **Logits:** raw class scores before sigmoid or softmax.
- **One-hot label:** vector with one 1 for the correct class and 0 for others.
- **Batch loss:** loss calculated over a group of examples.

```python
import numpy as np

y_true = np.array([1, 0, 1, 1])
y_pred = np.array([0.9, 0.2, 0.4, 0.8])

mse = np.mean((y_true - y_pred) ** 2)
bce = -np.mean(y_true * np.log(y_pred) + (1 - y_true) * np.log(1 - y_pred))
print("MSE:", round(mse, 4))
print("Binary cross entropy:", round(bce, 4))

class_true = np.array([0, 1, 0])
class_pred = np.array([0.1, 0.8, 0.1])
cce = -np.sum(class_true * np.log(class_pred))
print("Categorical cross entropy:", round(cce, 4))
```
