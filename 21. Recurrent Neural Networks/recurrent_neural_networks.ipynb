# Recurrent Neural Networks

A beginner-friendly introduction to recurrent neural networks.


## Learning Goals

- Understand sequence modeling
- Explain hidden state
- Run a tiny sequence example


## Concept Notes

RNNs process one time step at a time while carrying a hidden state that summarizes earlier steps.


## Step-by-Step Example

Run this compact example first. Focus on the inputs, the transformation, and the output.


```python
import numpy as np

sequence = np.array([1, 2, 3, 4])
hidden = 0
weight_input = 0.5
weight_hidden = 0.8

for value in sequence:
    hidden = np.tanh(weight_input * value + weight_hidden * hidden)
    print("hidden state:", round(hidden, 3))
```
