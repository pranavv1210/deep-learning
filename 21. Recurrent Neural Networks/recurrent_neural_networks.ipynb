# Recurrent Neural Networks

A beginner-friendly introduction to recurrent neural networks.


## Recurrent Neural Networks

RNNs process sequences one step at a time while carrying a hidden state that summarizes earlier context. This makes them suitable for ordered data such as text, sensor readings and time series.

## Learning Goals

- Understand sequence modeling
- Explain hidden state
- Run a tiny sequence example


### Hidden state and sequence length

At each time step the network combines the current input with the previous hidden state to produce a new state and possibly an output. The hidden state is a learned representation of recent sequence context, not a perfect memory of the entire past.

## Concept Notes

RNNs process one time step at a time while carrying a hidden state that summarizes earlier steps.


### Backpropagation through time

Training unfolds the recurrent computation across time steps and applies backpropagation through that unrolled graph. Long sequences can lead to vanishing or exploding gradients, making it hard to learn dependencies far in the past.

## Step-by-Step Example

Run this compact example first. Focus on the inputs, the transformation, and the output.


## Key terms

- **Sequence:** ordered set of observations.
- **Time step:** one position in a sequence.
- **Hidden state:** internal representation passed between steps.
- **Unrolling:** representing recurrent operations across sequence positions.
- **Vanishing gradient:** gradients become extremely small through many steps.
- **Exploding gradient:** gradients become extremely large.

```python
import numpy as np

sequence = np.array([1, 2, 3, 4])
hidden = 0
weight_input = 0.5
weight_hidden = 0.8

for value in sequence:
    hidden = np.tanh(weight_input * value + weight_hidden * hidden)
    print("hidden state:", round(hidden, 3))
```
