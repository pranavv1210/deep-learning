# Backpropagation

A beginner-friendly introduction to backpropagation.


## What is backpropagation?

Backpropagation efficiently calculates how each trainable weight affects the final loss. It applies the chain rule from calculus, starting at the output and moving backward through the network. The optimizer then uses these gradients to update parameters.

## Learning Goals

- Use the chain rule
- Compute gradients for a tiny network
- Connect gradients to parameter updates


### A tiny chain-rule example

Suppose `u = 2*x` and `y = u*u`. The chain rule says `dy/dx = (dy/du) * (du/dx)`. Since dy/du = 2u and du/dx = 2, the derivative is 4u. Backpropagation repeats this idea through many operations in a computational graph.

## Concept Notes

Backpropagation applies the chain rule from the loss backward through the network so each weight receives a useful update.


### Forward values and backward gradients

The forward pass stores intermediate values needed for derivatives. The backward pass combines local derivatives to calculate gradients with respect to earlier parameters. This is why frameworks build a computational graph and track operations when automatic differentiation is enabled.

## Step-by-Step Example

Run this compact example first. Focus on the inputs, the transformation, and the output.


## Key terms

- **Chain rule:** derivative rule for composed functions.
- **Computational graph:** representation of calculations and dependencies.
- **Gradient:** derivative of loss with respect to a parameter.
- **Automatic differentiation:** software technique that applies derivative rules to a computation.
- **Backward pass:** propagation of gradients from output toward inputs.

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

x = 2.0
w = 3.0
b = 1.0
y_true = 10.0

y_pred = w * x + b
loss = (y_pred - y_true) ** 2

d_loss_d_pred = 2 * (y_pred - y_true)
d_pred_d_w = x
d_pred_d_b = 1

d_loss_d_w = d_loss_d_pred * d_pred_d_w
d_loss_d_b = d_loss_d_pred * d_pred_d_b

print("prediction:", y_pred, "loss:", loss)
print("gradient w:", d_loss_d_w, "gradient b:", d_loss_d_b)
```
