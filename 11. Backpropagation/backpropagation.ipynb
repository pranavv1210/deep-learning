# Backpropagation

A beginner-friendly introduction to backpropagation.


## Learning Goals

- Use the chain rule
- Compute gradients for a tiny network
- Connect gradients to parameter updates


## Concept Notes

Backpropagation applies the chain rule from the loss backward through the network so each weight receives a useful update.


## Step-by-Step Example

Run this compact example first. Focus on the inputs, the transformation, and the output.


```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

x = 2.0
w = 3.0
b = 1.0
y_true = 10.0

y_pred = w * x + b
loss = (y_pred - y_true) ** 2

d_loss_d_pred = 2 * (y_pred - y_true)
d_pred_d_w = x
d_pred_d_b = 1

d_loss_d_w = d_loss_d_pred * d_pred_d_w
d_loss_d_b = d_loss_d_pred * d_pred_d_b

print("prediction:", y_pred, "loss:", loss)
print("gradient w:", d_loss_d_w, "gradient b:", d_loss_d_b)
```
