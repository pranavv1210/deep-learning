# Tensor Basics

Understand tensors, shapes, dtypes, and operations.


## What is a tensor?

A tensor is a container for numbers arranged along zero or more dimensions. A scalar has no axes, a vector has one axis, a matrix has two axes, and an image batch commonly has four axes. Deep-learning frameworks use tensors for input data, weights, intermediate activations and gradients.

## Learning Goals

- Represent scalars, vectors, matrices, and higher-dimensional tensors
- Inspect tensor shapes and dtypes
- Perform basic tensor operations


### Read shape and dtype

The **shape** tells how many entries exist along each axis. The **dtype** tells how values are represented in memory, such as int32 or float32. A batch shape `(8, 28, 28, 1)` means 8 images, each 28 pixels high, 28 pixels wide and with 1 channel. Some frameworks use channels-first `(batch, channels, height, width)` instead; always check the expected convention.

## Concept Notes

A tensor is a multi-dimensional array. Deep learning frameworks use tensors to store inputs, weights, activations, and gradients.


### NumPy arrays as tensors

NumPy is useful for learning tensor shapes without involving a deep-learning framework. `np.zeros` creates an array filled with zeros. The trailing `1` in the image shape is the grayscale channel count. Color RGB images commonly use 3 channels.

## NumPy Tensor Examples

The idea of a tensor can be introduced with NumPy arrays.


### Matrix multiplication and broadcasting

`x @ w` performs matrix multiplication. Here the two rows of x are examples and the two entries in each row are input features. The weight matrix maps two input features to one output value. Adding `b` uses broadcasting to add the same bias to each row. The result contains one output per example.

```python
import numpy as np

scalar = np.array(7)
vector = np.array([1, 2, 3])
matrix = np.array([[1, 2], [3, 4]])
image_batch = np.zeros((8, 28, 28, 1))

for name, value in [("scalar", scalar), ("vector", vector), ("matrix", matrix), ("image_batch", image_batch)]:
    print(name, "shape:", value.shape, "dtype:", value.dtype)
```

## Tensor Operations

Most model computation is built from addition, multiplication, matrix multiplication, and reshaping.


## Key terms and shape checklist

- **Axis/dimension:** one direction of a tensor's shape.
- **Batch:** group of examples processed together.
- **Channel:** image component such as red, green, blue, or grayscale.
- **Dtype:** numeric storage type.
- **Broadcasting:** rules that allow compatible shapes to participate in one operation.
- **Matrix multiplication:** combines dimensions according to linear algebra rules.

Before every operation, write down the input shapes, the expected output shape and whether the framework uses channels-first or channels-last.

```python
x = np.array([[1.0, 2.0], [3.0, 4.0]])
w = np.array([[0.2], [0.8]])
b = 0.5

output = x @ w + b
print(output)
```
