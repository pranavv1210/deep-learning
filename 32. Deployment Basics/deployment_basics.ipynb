# Deployment Basics

A beginner-friendly introduction to deployment basics.


## What does deployment mean?

Deployment makes a trained model available to an application or user. It may run in a web API, batch job, mobile app or edge device. Deployment includes more than loading weights: input validation, preprocessing, output handling and monitoring matter too.

## Learning Goals

- Separate training from inference
- Create a prediction function
- Understand simple API deployment flow


## Typical prediction API flow

A client sends input data to an endpoint. The service validates the schema, applies the same preprocessing used during training, calls the model and returns a prediction in a stable response format. Invalid inputs and model errors should be handled deliberately.

## Concept Notes

Deployment means making a trained model available for predictions. Start by writing a reliable inference function.


## Monitor deployed models

Monitor latency, errors, input distributions and prediction distributions. When labels become available, monitor model quality too. **Data drift** means incoming input distributions change; **concept drift** means the relationship between inputs and target changes.

## Step-by-Step Example

Run this compact example first. Focus on the inputs, the transformation, and the output.


## Key terms

- **Endpoint:** network address where a service accepts requests.
- **Inference service:** application that runs a trained model for predictions.
- **Latency:** time taken to respond.
- **Data drift:** change in incoming input distribution.
- **Concept drift:** change in the relationship the model is trying to learn.
- **Monitoring:** ongoing measurement of service behavior and model quality.

```python
import joblib
import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn.datasets import make_classification

X, y = make_classification(n_samples=200, n_features=4, random_state=42)
model = LogisticRegression(random_state=42).fit(X, y)
joblib.dump(model, "demo_model.joblib")

def predict_one(features):
    loaded_model = joblib.load("demo_model.joblib")
    features = np.array(features).reshape(1, -1)
    return int(loaded_model.predict(features)[0])

print(predict_one(X[0]))
```
